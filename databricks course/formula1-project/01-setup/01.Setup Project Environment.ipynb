#Set-up the project environment for Formula 1 project
1. create external location databricks-course-ext-dl101-formula1
2. create catalog formula1
3. create schemas landing, bronze, silver and gold
4. create volume files in the landing schema


##Access cloud storage


```sql
%fs ls 'abfss://formula1@databrickscourseextdl010.dfs.core.windows.net/landing'
```

##Create external location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_course_ext_dl010_formula1
URL 'abfss://formula1@databrickscourseextdl010.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'external location for the formula1 container';
```

##create catalog formula 1

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS formula1
  MANAGED LOCATION 'abfss://formula1@databrickscourseextdl010.dfs.core.windows.net/'
  COMMENT 'This is a main catalog for the formula1 project';
```

##create schemas landing, bronze, silver and gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
  MANAGED LOCATION 'abfss://formula1@databrickscourseextdl010.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
  MANAGED LOCATION 'abfss://formula1@databrickscourseextdl010.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1.gold
  MANAGED LOCATION 'abfss://formula1@databrickscourseextdl010.dfs.core.windows.net/gold';
```

```sql
SELECT current_catalog();
```

```sql
USE CATALOG formula1;

```

```sql
SHOW SCHEMAS;
```

##create volume files


```sql
CREATE EXTERNAL VOLUME formula1.landing.files
     LOCATION 'abfss://formula1@databrickscourseextdl010.dfs.core.windows.net/landing';
     
```
