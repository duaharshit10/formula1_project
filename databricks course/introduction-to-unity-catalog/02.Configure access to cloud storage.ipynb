### configure access to cloud storage using unity catalog 


### Access cloud storage 

```sql
%fs ls 'abfss://demo@databrickscourseextdl010.dfs.core.windows.net/'
```

### Create external location 


```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_course_ext_dl010_demo
URL 'abfss://demo@databrickscourseextdl010.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'external location for the demo container';
```
